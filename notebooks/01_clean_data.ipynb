```python
import pandas as pd
```

```python
df = pd.read_csv('goodreads_onestar.csv')
```

```python
df.head()
```

## DATA CLEANING

```python
len(df)
```

```python
# dropping rows with missing review_text

df_clean = df.dropna(subset=["review_text"]) 
```

```python
# importing the langdetect library to detect the language of the review_text column

from langdetect import detect, DetectorFactory

DetectorFactory.seed = 0

def safe_detect(t):
    try: return detect(t)
    except: return "unknown"
```

```python
# labeling the language of the review_text column and filtering for only English reviews

df_clean["lang"] = df_clean["review_text"].map(safe_detect)
df_clean = df_clean[df_clean["lang"] == "en"]
```

```python
# dropping duplicate reviews based on user_id and review_text

df_clean = df_clean.drop_duplicates(subset=["user_id", "review_text"])
```

```python
# cleaned data is saved to a new CSV file

df_clean.to_csv("goodreads_onestar_clean.csv", index=False)
```
